```python
import re
renamed_column = []
for col in df.columns:
    col=re.sub(r"[^a-zA-Z0-9]+","_",col)
    col=re.sub(r"(a-z0-9)(A-Z)",r"\1_\2",col)
    col = re.sub(r"_+", "_", col).strip("_").lower()
    renamed_column.append(col)
    print(col)
df.columns = renamed_column
    
```

```python
# check each numerical columns min and max value to convert it into appropriate datatypes
numerical_columns = df.select_dtypes(include='number').columns
numerical_columns
```

```python
# checking each numerical columns max and minimum values and compare it with above min and max to convert it into appropriate datatype
for col in numerical_columns:
    try:
        if (df[col].min() >= np.iinfo("int8").min and df[col].max() <= np.iinfo("int8").max) :
                                  df[col]= df[col].astype(np.int8)
        elif (df[col].min() >= np.iinfo("int16").min and df[col].max() <= np.iinfo("int16").max) :
                                   df[col]= df[col].astype(np.int16)
        elif (df[col].min() >= np.iinfo("int32").min and df[col].max() <= np.iinfo("int32").max):
                                   df[col]= df[col].astype(np.int32)
        elif (df[col].min() >= np.iinfo("int64").min and df[col].max() <= np.iinfo("int64").max) :
                                    df[col]=df[col].astype(np.int64)
    except (TypeError, ValueError, OverflowError) as error:
        print(f"Error while converting column '{col}': {error}")

        # Raise the same original error again.
        raise                                                          
```
