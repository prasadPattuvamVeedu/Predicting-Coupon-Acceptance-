```python
import re

def clean_column_names(columns):

    renamed_columns = []

    for col in columns:
        col = str(col)
        col = re.sub(r"([a-z0-9])([A-Z])", r"\1_\2", col)


        col = re.sub(r"[^a-zA-Z0-9]+", "_", col)


        col = re.sub(r"_+", "_", col).strip("_").lower()

        renamed_columns.append(col)

    return renamed_columns
df.columns = clean_column_names(df.columns)
```

```python
# check each numerical columns min and max value to convert it into appropriate datatypes
numerical_columns = df.select_dtypes(include='number').columns
numerical_columns
```

```python
# checking each numerical columns max and minimum values and compare it with above min and max to convert it into appropriate datatype
datatype_column = {}
for col in numerical_columns:
    try:
        if (df[col].min() >= np.iinfo("int8").min and df[col].max() <= np.iinfo("int8").max) :
                                  df[col]= df[col].astype(np.int8)
                                  datatype_column[col] = np.int8
        elif (df[col].min() >= np.iinfo("int16").min and df[col].max() <= np.iinfo("int16").max) :
                                   df[col]= df[col].astype(np.int16)
                                   datatype_column[col] = np.int16
        elif (df[col].min() >= np.iinfo("int32").min and df[col].max() <= np.iinfo("int32").max):
                                   df[col]= df[col].astype(np.int32)
                                   datatype_column[col] = np.int32
        elif (df[col].min() >= np.iinfo("int64").min and df[col].max() <= np.iinfo("int64").max) :
                                    df[col]=df[col].astype(np.int64)
                                    datatype_column[col] = np.int64
    except (TypeError, ValueError, OverflowError) as error:
        print(f"Error while converting column '{col}': {error}")

        # Raise the same original error again.
        raise                                                          
```

7 Perform hypothesis testing using Chi-square and Cramér's V
</h1>

```python
from scipy.stats import chi2_contingency
```

```python
significance_variable = {}
not_significance_variable ={}
cramers_V = {}
for col in df.iloc[:,:-1]:
    cross_tab = pd.crosstab(df[col],df.iloc[:,-1])
    ch2,p_value,dof,expected = chi2_contingency(cross_tab)
    
    if p_value <= 0.05:
        significance_variable[col] = p_value
    else:
        not_significance_variable[col]=p_value

    n = cross_tab.to_numpy().sum()
    r,c = cross_tab.shape
    value = np.sqrt(ch2/(n*(r-1)*(c-1)))
    if value <= 0.10:
        cramers_V[col] = [value,"negligible"]
    elif value <=.30:
        cramers_V[col] = [value,"weak"]
    elif value < 0.50:
        cramers_V[col] = [value,"moderate"]
    else:
        cramers_V[col] = [value,"strong"]
print(
    f"{'Significant features':<30}: {significance_variable}"
    f"\n\n"
    f"{'Not significant features':<30}: {not_significance_variable}"
)
print(f"---------------------------------------------------")
print("-" * 50)
print("Cramér's V Results")

for col, result in cramers_V.items():
    value, strength = result

    if strength.lower() == "negligible":
        print("negligible")
        print(f"{col:<25} Cramér's V = {value:.4f}")
    if strength.lower() == "weak":
        print("weak")
        print(f"{col:<25} Cramér's V = {value:.4f}")
    if strength.lower() == "moderate":
        print("moderate")
        print(f"{col:<25} Cramér's V = {value:.4f}")
    if strength.lower() == "strong":
        print("strong")
        print(f"{col:<25} Cramér's V = {value:.4f}")

        
```

<h4 style="color: #15803d; margin-top: 0;">
📊 Chi-Square and Cramér’s V Inference
</h4>

<p style="line-height: 1.7; margin-bottom: 10px;">
The Chi-square test shows that all tested features except
<strong>direction_same</strong> have p-values below <strong>0.05</strong>.
Therefore, the null hypothesis of independence is rejected for these features,
indicating that they have a statistically significant association with coupon
acceptance.
</p>

<p style="line-height: 1.7; margin-bottom: 10px;">
The <strong>direction_same</strong> feature has a p-value of
<strong>0.072</strong>, which is greater than 0.05. Therefore, the null
hypothesis cannot be rejected, and there is insufficient evidence to conclude
that direction_same is significantly associated with coupon acceptance.
</p>

<p style="line-height: 1.7; margin-bottom: 10px;">
Cramér’s V was calculated to measure the strength of these associations.
The features <strong>coupon</strong> (0.1314),
<strong>expiration</strong> (0.1282), and
<strong>tocoupon_geq25min</strong> (0.1076) show the highest association
strengths. However, these relationships are still classified as
<strong>weak</strong>.
</p>

<p style="line-height: 1.7; margin-bottom: 10px;">
All remaining features have Cramér’s V values below <strong>0.10</strong>,
indicating negligible individual associations with coupon acceptance.
The <strong>direction_same</strong> feature has the lowest Cramér’s V value
of <strong>0.0162</strong>, confirming that its individual relationship with
coupon acceptance is extremely small.
</p>

<p style="line-height: 1.7; margin-bottom: 0;">
<strong>Conclusion:</strong> Although most features are statistically
significant according to the Chi-square test, their association strengths are
weak or negligible. This difference may occur because a large dataset can
produce very small p-values even when the practical relationship is small.
Therefore, features should not be removed solely based on Chi-square p-values
or Cramér’s V values. Their business relevance and contribution to model
performance should also be evaluated.
</p>

    <strong>Perturbation Test:</strong>
    A small amount of random noise is added to the input data, and the model is retrained.
    Large changes in coefficients may indicate multicollinearity or model instability.
</div>

```python
from sklearn.linear_model import LogisticRegression
```

```python
base_model_coef={}
lr = LogisticRegression()
model_lr = lr.fit(x_train_sc,y_train)
coefficients=model_lr.coef_.ravel()
base_model_coef={col:coeff for col,coeff in  zip(x_train.columns,coefficients)}
```

```python
base_coef=pd.DataFrame(base_model_coef.items(),columns=["Feature","coefficient"])
base_coef
```

```python
# Adding small impurties in train model x_train_sc
np.random.seed(42)
impurity = np.random.normal(loc=0.001,scale=0.01,size=x_train_sc.shape)
```

```python
x_train_sc_pertubated = x_train_sc+impurity
```

```python
pert_model_coef={}
lr = LogisticRegression()
model_lr_pert = lr.fit(x_train_sc_pertubated,y_train)
coefficients_pert=model_lr_pert.coef_.ravel()
pert_model_coef={col:coeff for col,coeff in  zip(x_train.columns,coefficients_pert)}
```

```python
pert_coef = pd.DataFrame(pert_model_coef.items(),columns=["Feature","pert_coefficient"])
coeff_df = base_coef.merge(pert_coef,on="Feature")
```

```python
coeff_df
```

```python
coeff_df["percentage_change"] = coeff_df.apply(lambda x : abs((x["pert_coefficient"]-x["coefficient"])/x["coefficient"])*100,axis=1)
```

<h4 style="color: #1d4ed8; margin-top: 0;">
📊 Scaled and Unscaled Dataset Usage
</h4>

<p style="line-height: 1.7;">
Two versions of the same training and testing dataset are maintained during
model building:
</p>

<ul style="line-height: 1.8;">
    <li>
        <strong>Scaled dataset:</strong> Used for models that are sensitive to
        feature magnitude, such as Logistic Regression, K-Nearest Neighbors,
        and Support Vector Machine.
    </li>
    <li>
        <strong>Unscaled dataset:</strong> Used for tree-based models such as
        Decision Tree, Random Forest, and XGBoost.
    </li>
</ul>

<p style="line-height: 1.7; margin-bottom: 0;">
Decision Tree, Random Forest, and XGBoost do not require feature scaling because
they split the data using feature thresholds. Scaling changes the numerical
range of the features but does not change their order. Therefore, the original
unscaled training and testing datasets are used for these models.
</p>

</div>


```python
# for scaled
from sklearn.linear_model import LogisticRegression
from sklearn.svm import SVC
from sklearn.neighbors import KNeighborsClassifier
from sklearn.naive_bayes import GaussianNB

# for unscaled
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier
from xgboost import XGBClassifier
```

```python
required_scale_model = ["Logistic Regression","Support Vector Classifier","K-Nearest Neighbors","Gaussian Naive Bayes"]
```

```python
from sklearn.model_selection import GridSearchCV, StratifiedKFold
```

```python
model_detail = [
    {
        "name": "Logistic Regression",
        "model": LogisticRegression(
            max_iter=2000,
            random_state=42
        ),
        "parameter": [
            {
                "solver": ["liblinear"],
                "penalty": ["l1", "l2"],
                "C": [0.1, 1, 10]
            },
            {
                "solver": ["lbfgs"],
                "penalty": ["l2"],
                "C": [0.1, 1, 10]
            }
        ],
        "scaled": True
    },

    {
        "name": "Support Vector Classifier",
        "model": SVC(
            cache_size=2000,
            random_state=42
        ),
        "parameter": [
            {
                # Gamma is not required for the linear kernel.
                "kernel": ["linear"],
                "C": [0.1, 1, 10]
            },
            {
                "kernel": ["rbf"],
                "C": [0.1, 1, 10],
                "gamma": ["scale", 0.1]
            }
        ],
        "scaled": True
    },

    {
        "name": "K-Nearest Neighbors",
        "model": KNeighborsClassifier(),
        "parameter": {
            "n_neighbors": [5, 9, 15],
            "weights": ["uniform", "distance"],
            "metric": ["euclidean", "manhattan"]
        },
        "scaled": True
    },

    {
        "name": "Gaussian Naive Bayes",
        "model": GaussianNB(),
        "parameter": {
            "var_smoothing": [1e-10, 1e-9, 1e-8]
        },
        "scaled": False
    },

    {
        "name": "Decision Tree",
        "model": DecisionTreeClassifier(
            random_state=42
        ),
        "parameter": {
            "criterion": ["gini", "entropy"],
            "max_depth": [5, 10, None],
            "min_samples_split": [2, 10],
            "min_samples_leaf": [1, 5]
        },
        "scaled": False
    },

    {
        "name": "Random Forest",
        "model": RandomForestClassifier(
            random_state=42,
            n_jobs=-1
        ),
        "parameter": {
            "n_estimators": [100, 200],
            "max_depth": [10, None],
            "min_samples_split": [2, 10],
            "max_features": ["sqrt"]
        },
        "scaled": False
    },

    {
        "name": "XGBoost",
        "model": XGBClassifier(
            random_state=42,
            eval_metric="logloss",
            n_jobs=-1
        ),
        "parameter": {
            "n_estimators": [100, 200],
            "max_depth": [3, 5],
            "learning_rate": [0.05, 0.1]
        },
        "scaled": False
    }
]
```

```python
best_estimator = {}
for model in model_detail:
    print(f"{'Model selection':<20}: {model['name']}")
    if model["name"] in required_scale_model:
        gr = GridSearchCV(estimator=model["model"],param_grid=model["parameter"], scoring="accuracy",cv=5,n_jobs=-1)
        gr.fit(x_train_sc,y_train)
        best_estimator[model["name"]] = {"estimator":gr.best_estimator_}
    else:
        gr = GridSearchCV(estimator=model["model"],param_grid=model["parameter"], scoring="accuracy",cv=5,n_jobs=-1)
        gr.fit(x_train,y_train)
        best_estimator[model["name"]] = {"estimator":gr.best_estimator_}
```

```python
best_estimator
```

```python
prediction={}
for name,est in best_estimator.items():
   print(name)
   print(est)
   
   for _,model_est in est.items():
       if name in required_scale_model:
             prediction[name]=model_est.predict(x_test_sc)
       else:
               prediction[name] =model_est.predict(x_test)
           
```

```python
df_prediction=pd.DataFrame(prediction)
```
